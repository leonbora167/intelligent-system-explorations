```python
from ddgs import DDGS
from trafilatura import fetch_url, extract, sitemaps
from trafilatura.spider import focused_crawler
import requests 
from bs4 import BeautifulSoup
from playwright.sync_api import sync_playwright
```

# Link Retrieval

```python
#Textual retrieval
query = "Metal Gear Solid"

results = DDGS().text(query, 
                      safesearch = "off",
                      backend = "auto",
                      max_results=5)
```

```python
results
```

# Read/Extract data from link

```python
web_url = results[0]["href"]
downloaded = fetch_url(web_url)
print(web_url)
```

```python
#plain text output 

result = extract(downloaded)
print(result[0:100])
```

```python
# XML Format, to preserver doc structure ?
xml_result = extract(downloaded, output_format="xml")

# Json format
json_result = extract(downloaded, output_format="json", with_metadata=True, include_comments=True)

# Markdown format 
markdown_result = extract(downloaded, output_format="markdown", with_metadata=True)
```

# BS4 Parser

```python
html_content_file = "page_content.html"
with open(html_content_file, "r", encoding="utf-8") as f:
    html_content = f.read() 
soup = BeautifulSoup(html_content, "html.parser")

page_main_title = soup.title.string
all_paragraphs = soup.find_all("p")
for index, paragraph in enumerate(all_paragraphs):
    print(str(index) + " : " + paragraph.text)
    break
```

```python
print("Main title is : ", page_main_title)

all_links = soup.find_all("a")
for index, link in enumerate(all_links):
    url = link.get("href")
    if "https" in url:
        print(str(index) + " : " + link.get('href'))
        break
```
