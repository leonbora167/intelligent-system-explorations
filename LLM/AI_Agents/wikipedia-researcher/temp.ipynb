```python
from ddgs import DDGS
from trafilatura import fetch_url, extract, sitemaps
from trafilatura.spider import focused_crawler
import requests 
from bs4 import BeautifulSoup
```

# Link Retrieval

```python
#Textual retrieval
query = "Lego Batman 2026 game"

results = DDGS().text(query, 
                      safesearch = "off",
                      backend = "auto",
                      max_results=5)
```

```python
results
```

# Read/Extract data from link

```python
web_url = results[0]["href"]
downloaded = fetch_url(web_url)
```

```python
#plain text output 

result = extract(downloaded)
print(result[0:100])
```

```python
# XML Format, to preserver doc structure ?
result = extract(downloaded, output_format="xml")

# Json format
result = extract(downloaded, output_format="json", with_metadata=True, include_comments=True)

# Markdown format 
result = extract(downloaded, output_format="markdown", with_metadata=True)
```

# Extract links within sites

```python
html = requests.get(web_url).text 
soup = BeautifulSoup(html, "lxml")

links = []

for a in soup.find_all("a", href=True):
    links.append(a["href"])

print(links)
```
